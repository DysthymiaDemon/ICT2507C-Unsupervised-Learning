```python
missing_percent = df.isnull().sum() / len(df) * 100
print(missing_percent)
```

```python
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
df.head()
```

```python
df['Description'] = df['Description'].str.strip()
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
df['InvoiceNo']=df['InvoiceNo'].astype('str')
df = df[-df['InvoiceNo'].str.contains('C')]
df
```

```python
basket = (df[df['Country'] =="France"]
         .groupby(['InvoiceNo', 'Description'])['Quantity']
         .sum().unstack().reset_index().fillna(0)
         .set_index('InvoiceNo'))
basket
```

```python
def encode(a):
    if a<= 0:
        return 0
    if a>= 1:
        return 1

basketset = basket.applymap(encode)
basketset.drop('POSTAGE', inplace=True, axis=1)
basketset
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules


frequentitem = apriori(basketset, min_support=0.05, use_colnames=True)
rules = association_rules(frequentitem, metric="lift",min_threshold=1)
rules.head()
```

```python
rules[(rules['lift'] >= 6) & (rules['confidence'] >= 0.8)]
```

 Interpretation:

Customers buying "Alarm Clock Bakelike Green" have a very high chance (81.6%) of also buying "Alarm Clock Bakelike Red." Similarly, the reverse holds.

"Childrens Cutlery Dolly Girl" and "Childrens Cutlery Spaceboy" have an even stronger correlation, meaning if one is purchased, the other is almost always bought.


Lift values are significantly higher than 1, confirming these are strong, non-random associations.
