## Task 1 - Answer the question
- What is Market Basket Analysis? Provide an example where this is often used.

Market Basket Analysis (MBA), also known as association rule mining or affinity analysis, is a data mining technique used to discover patterns, relationships, and associations among items or products purchased together in a transaction or shopping basket.

Let's consider a supermarket as an example. In a supermarket, Market Basket Analysis can be applied as follows:

- Discovering Product Associations: By analyzing transaction data, the supermarket may find that customers who purchase bread are likely to buy butter as well (association rule: {Bread} -> {Butter}).
- Cross-Selling: Based on the discovered rules, the supermarket can place bread and butter on nearby shelves or offer discounts when both items are purchased together.

- Perform market basket analysis on the dataset using the apriori algorithm API from the apyori library package.

```python
import numpy as np
import pandas as pd
import seaborn as sns
from apyori import apriori
```

```python
missing_percent = df.isnull().sum() / len(df) * 100
print(missing_percent)
```

```python
df_cleaned = df.dropna(axis=1, thresh=len(df) * 0.3)  # Keep columns with at least 30% non-null values
```

```python
print(df_cleaned.isnull().sum())
```

```python
missing_percent = df_cleaned.isnull().sum() / len(df_cleaned) * 100
print(missing_percent)
```

```python
# For categorical columns (impute with mode)
df_filled = df_cleaned.apply(lambda x: x.fillna(x.mode()[0]) if x.dtype == "object" else x)
```

```python
df_filled
```

```python
# convert the dataframe into list
transactions = []

for i in range(len(df_filled)):
    transaction = []
    for j in range(5):
        transaction.append(str(df_filled.iloc[i, j]))
    transactions.append(transaction)
```

```python
transactions[0]
```

```python
# analyze using apriori algorithm

association_rules = apriori(transactions, min_support=0.005, min_confidence=0.3, min_lift=3, min_length=2)
results = list(association_rules)
```

```python
results
```

```python
# Function to extract rules from results
def inspect(results):
    rule_list = []
    for rule in results:
        pair = rule.items  # Base item + associated item
        items = [x for x in pair]
        
        # Rule details
        support = rule.support
        for ordered_stat in rule.ordered_statistics:
            base = list(ordered_stat.items_base)
            add = list(ordered_stat.items_add)
            confidence = ordered_stat.confidence
            lift = ordered_stat.lift
            rule_list.append([base, add, support, confidence, lift])
    
    return pd.DataFrame(rule_list, columns=["Base Item", "Added Item", "Support", "Confidence", "Lift"])

# Convert rules into DataFrame
rules_df = inspect(results)

# Display top 10 rules sorted by lift
rules_df.sort_values(by="Lift", ascending=False).head()
```
