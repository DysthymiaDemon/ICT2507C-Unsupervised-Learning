```python
import numpy as np
import pandas as pd
import seaborn as sns
# from apyori import apriori
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_excel('Online Retail.xlsx')
df.head(10)
```

```python
missing_percent = df.isnull().sum() / len(df) * 100
print(missing_percent)
```

```python
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
df.head()
```

```python
df['Description'] = df['Description'].str.strip()
```

```python
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
```

```python
df['InvoiceNo']=df['InvoiceNo'].astype('str')
```

```python
df = df[-df['InvoiceNo'].str.contains('C')]
df
```

```python
basket = (df[df['Country'] =="France"]
         .groupby(['InvoiceNo', 'Description'])['Quantity']
         .sum().unstack().reset_index().fillna(0)
         .set_index('InvoiceNo'))
basket
```

```python
def encode(a):
    if a<= 0:
        return 0
    if a>= 1:
        return 1

basketset = basket.applymap(encode)
basketset.drop('POSTAGE', inplace=True, axis=1)
basketset
```

```python
# analyze using apriori algorithm

frequentlybought = apriori(basketset, min_support=0.1, use_colnames=True)
rules = association_rules(frequentlybought, metric="lift",min_threshold=1)
rules.head()
```

```python
rules[(rules['lift'] >= 3) & (rules['confidence'] >= 0.6)]
```

```python
rules.sort_values(by="lift", ascending=False).head(20)
```

```python
rules.sort_values(by="lift", ascending=True).head(20)
```
